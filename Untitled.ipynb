```python
import numpy as np
import matplotlib.pyplot as plt
oscar_movies=[
    "The dark Knight",
    "The hut locker",
    "The king's speech",
    "The artist",
    "Argo"
]

oscar_revenue=[1005,170,427,133,232]

non_oscar_movies=["dhamaal4","bahubali","spider man","lalla"]
non_oscar_revenue=[378,2788,829,185,275]
years=[2008,2009,2010,2011,2012]

x = np.arange(len(years))


width=0.4
plt.bar(x-width/2,oscar_revenue,width, label="Oscar Movies")
plt.bar(x+width/2,non_oscar_revenue,width, label="Non-Oscar movies")

plt.title("Oscar vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue($M)")
plt.legend()
```

```python
#subplots
import numpy as np
import matplotlib.pyplot as plt
x=[1,2,3,5,6,7,8,9,10]
y1=[np.sqrt(i) for i in x] #square root
y2=[i*2 for i in x]
y3=[i**2 for i in x]
y4=[i**3 for i in x]

plt.subplot(2,2,1)
plt.plot(x,y1)
plt.title("Plot1-square root")

plt.subplot(2,2,2)
plt.scatter(x,y2)
plt.title("Plot2 - double ")

plt.subplot(2,2,3)
plt.plot(x,y3)
plt.title("Plot3 - square")

plt.subplot(2,2,4)
plt.plot(x,y4)
plt.title("Plot4 - cube ")

plt.tight_layout()
```
