```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.get_dataset_names()
```

```python
tips=sns.load_dataset("tips")
```

```python
tips
```

```python
print(type(tips))
```

```python
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    showmeans=True,
    hue="sex"
)
```

```python
tips
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
)
plt.title("Total bill vs Tip")
plt.xlabel("total_bill")
plt.ylabel("Tip")
```
